```python
import numpy as np
import matplotlib.pyplot as plt
weight1000 = np.random.normal(loc=50, scale=0.1, size=1000)
plt.hist(weight1000, edgecolor='black')
plt.show() #
```

```python
import seaborn as sns
#sns.histplot(weight1000, bins=30)
sns.histplot(weight1000, bins=30, binrange=(min(weight1000), max(weight1000)))
```
